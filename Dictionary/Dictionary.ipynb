```python
#syntax
# syntax
empty_dict = {}
# Dictionary with data values
dct = {'key1':'value1', 'key2':'value2', 'key3':'value3', 'key4':'value4'}
```

```python
person ={
    "First Name" : "Muhammad",
    "Last Name" : "Azlan",
    "Age" : 25,
    "Gender" : "Male",
    "is_Married":False,
    "Country" : "India",
    "Skills" :["Python","C++","Java","SQL"],
    "Address": {
        "House Name": "kjsnbvkl",
        "street Name" : "wjdbjsnc",
        "Taluq" : "bvamnf",
        "pincode" :"76786878",
    }
}
print(person)
```

```python
print(len(person))
```

```python
#accessing dictionary by key
print(person["First Name"])
print(person["Skills"][0])
print(person["Address"]["Taluq"])
```

```python
print(person.get("First Name"))
print(person.get("Skills"))
print(person.get("Skills")[0])
print(person.get("Address"))
#print(person.get("Address)
```

```python
print(person.get("Address").get("street Name"))
```

```python
#adding items to dictionary
person["Job Title"] ="Software Engineer"
person["Skills"].append(".NET")
print(person)
```

```python
#modifying existing item in dictionary
person["First Name"] = "Yosuf"
person["Age"] = 24
print(person)
```

```python
#checking keys in dictionary
print("First Name" in person)
print("job" in person)
```

```python
#removing pop(),popitem(),del
person.pop("Gender")
print(person)
person.popitem()
print(person)
del person
#print(person)
```

```python
#changing dictionary to list
person ={
    "First Name" : "Muhammad",
    "Last Name" : "Azlan",
    "Age" : 25,
    "Gender" : "Male",
    "is_Married":False,
    "Country" : "India",
    "Skills" :["Python","C++","Java","SQL"],
    "Address": {
        "House Name": "kjsnbvkl",
        "street Name" : "wjdbjsnc",
        "Taluq" : "bvamnf",
        "pincode" :"76786878",
    }
}
print(person.items())
```

```python
#copy dictionary
copy_person=person.copy()
print(copy_person)
```

```python
#Getting Dictionary keys as a List
print(person.keys())
```

```python
#Getting Dictionary Values as a List
print(person.values())
```
