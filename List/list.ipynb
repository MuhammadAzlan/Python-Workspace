```python
#joining list
positive_num = [1,2,3,4,5,6]
zero = [0]
negative_num = [-1,-2,-3,-4,-5,-6]
print(positive_num + zero + negative_num)
fruits = ['banana', 'orange', 'mango', 'lemon']
vegetables = ['Tomato', 'Potato', 'Cabbage', 'Onion', 'Carrot']
fruits_and_vegetables = fruits + vegetables
print(fruits_and_vegetables )
```

```python
#joining values using extend() method
num1 = [1,2,3,4,5]
num2= [ 6,7,8,9,10]
num1.extend(num2)
print(num1)
```

```python
#counting items in al list
fruits = ['banana', 'orange', 'mango', 'lemon']
print(fruits.count("banana"))
numbers = [22,23,24,25,23,26,27,24,28]
print(numbers.count(23))
```

```python
#index() method returns index of the item in the list
fruits = ['banana', 'orange', 'mango', 'lemon']
print(fruits.index("mango"))
numbers = [22,23,24,25,23,26,27,24,28]
print(numbers.index(23))
```

```python
#sorting list irems
fruits = ['banana', 'orange', 'mango', 'lemon']
fruits.sort()
print(fruits)
fruits.sort(reverse=False)
print(fruits)
numbers = [22,23,24,25,23,26,27,24,28]
numbers.sort()
print(numbers)
numbers.sort(reverse=True)
print(numbers)
```

```python
fruits = ['banana', 'orange', 'mango', 'lemon']
print(sorted(fruits))   # ['banana', 'lemon', 'mango', 'orange']
# Reverse order
fruits = ['banana', 'orange', 'mango', 'lemon']
fruits = sorted(fruits,reverse=True)
print(fruits)     # ['orange', 'mango', 'lemon', 'banana']
```
