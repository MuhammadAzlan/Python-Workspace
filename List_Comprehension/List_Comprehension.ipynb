# **List Comprehension**

```python
# syntax
#[expression for i in iterable if condition]
```

For instance if you want to change a string to a list of characters. You can use a couple of methods. Let's see some of them:

```python
Language = "Python"
lst = list(Language)
print(type(lst))
print(lst)

lst = [i for i in Language]
print(type(lst))
print(lst)
```

```python
Job = "Software Engineer"
lst = [i for i in Job]
print(type(lst))
print(lst)
```

```python
numbers = [i for i in range(10)]
print(numbers)
```

```python
Squares =[i for i in range(11)]
print(Squares)
```

```python
num_sq = [(i,i*i) for i in range(11)]
print(num_sq)
```

```python
#List comprehension can be combined with if expression

numbers = [i for i in range(11) if i%2 ==0]
print(numbers)
```

```python
numbers = [i for i in range(11) if i%2 !=0]
print(numbers)
```

```python
# Filter numbers: let's filter out positive even numbers from the list below
numbers = [-8, -7, -3, -1, 0, 1, 3, 4, 5, 7, 6, 8, 10]
fliter_num = [i for i in numbers if i%2 ==0 and i>0]
print(fliter_num)
```

```python
# Flattening a two dimensional array
list_of_lists = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flattening_List = [number for row in list_of_lists for number in row]
print(flattening_List)
```

# **Lambda Function**

```python
# syntax
# x = lambda param1, param2, param3: param1 + param2 + param3
# print(x(arg1, arg2, arg3))
```

```python
#normal way
def add_num(a,b):
    return a+b
print(add_num(a=2,b=5))
```

```python
#lambda way
x = lambda a,b : a+b
print(x(a=2,b=5))
```

```python
# Self invoking lambda function
(lambda a,b : a+b) (2,5)
```

```python
square = lambda x : x ** 2
print(square(3))
```

```python
cube = lambda x : pow(x,3)
print(cube(7))
```

```python
multiple_variable = lambda a, b, c: a ** 2 - 3 * b + 4 * c
print(multiple_variable(5, 5, 3))
```

**Lambda Function Inside Another Function**

```python
def power(x):
    return lambda n : x**n

cube = power(2)(3)
print(cube)
two_power_of_five = power(2)(5)
print(two_power_of_five)
```
